# Path Manipulation:

### Splitting paths:

os.path.split(path): Splits the path into a tuple containing the directory and the file name.

```python
directory, file_name = os.path.split("/path/to/directory/file.txt")
print(directory)   # Output: /path/to/directory
print(file_name)   # Output: file.txt
```

### Getting the file extension:

os.path.splitext(path): Splits the path into a tuple containing the file name and the extension.

```python
file_name, extension = os.path.splitext("/path/to/file.txt")
print(file_name)   # Output: /path/to/file
print(extension)   # Output: .txt
```

## os.walk
#### When working with a large number of files and directories, it's often necessary to traverse through them and perform specific operations. Python's os module provides a convenient function called os.walk() that allows you to recursively walk through a directory tree and process files and subdirectories.
Understanding os.walk():

os.walk(directory) is a generator function that yields a tuple of three values for each directory it visits:
- The current directory path
- A list of subdirectories in the current directory
- A list of files in the current directory



It recursively traverses the directory tree, visiting each subdirectory and its contents.
Basic usage of os.walk():

```python
import os

directory = "lfw1"
for root, dirs, files in os.walk(directory):
    print("Current directory:", root)
    print("Subdirectories:", dirs)
    print("Files:", files)
    print("---")
```

```python
import os

def get_image_paths(directory):
    image_paths = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".jpg") or file.endswith(".png"):  # Adjust the file extensions as needed
                image_paths.append(os.path.join(root, file))
    return image_paths

# Example usage:
directory = "lfw1"  # Replace "lfw1" with the path to your desired directory
image_paths = get_image_paths(directory)
for path in image_paths:
    print(path)
```

## method

```python
import os

directory = "lfw1"
image_extensions = [".jpg", ".jpeg", ".png", ".gif"]

for root, dirs, files in os.walk(directory):
    for file in files:
        if any(file.lower().endswith(ext) for ext in image_extensions):
            image_path = os.path.join(root, file)
            print("Image found:", image_path)
```
