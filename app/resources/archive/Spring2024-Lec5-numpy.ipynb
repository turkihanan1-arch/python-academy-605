# Python Numpy


###### NumPy (Numerical Python) is a fundamental library for scientific computing in Python. It provides a powerful array object and a wide range of mathematical functions to operate on these arrays efficiently. NumPy is the foundation upon which many other scientific and data analysis libraries, such as SciPy, Pandas, and Matplotlib, are built.

- The core feature of NumPy is the ndarray (n-dimensional array) object, which allows you to store and manipulate large arrays of homogeneous data types. These arrays can be one-dimensional (1D), two-dimensional (2D), or of higher dimensions. NumPy arrays are more efficient and faster compared to Python's built-in lists, especially when working with large datasets.

- NumPy provides a wide range of functions and operators for performing mathematical operations on arrays, such as element-wise addition, subtraction, multiplication, and division. It also includes functions for more advanced operations like linear algebra, Fourier transforms, and random number generation.

#### Some key features and benefits of NumPy include:

- Efficient memory usage: NumPy arrays are stored in contiguous memory blocks, enabling faster access and manipulation compared to Python lists.

- fast as compared to the python List (30 times faster)
*  It has better support for mathematical operations.

- Vectorized operations: NumPy allows you to perform element-wise operations on arrays without the need for explicit loops, which leads to concise and readable code.

- Broadcasting: NumPy supports broadcasting, which enables arrays with different shapes to be used in arithmetic operations without the need for explicit reshaping.

- Integration with other libraries: Many scientific and data analysis libraries in Python are built on top of NumPy, making it easy to work with NumPy arrays across different libraries.

- Extensive mathematical functions: NumPy provides a comprehensive set of mathematical functions, including trigonometric functions, exponential functions, statistical functions, and more.

###### NumPy has become an essential tool in the Python ecosystem for various fields, including scientific computing, data analysis, machine learning, and more. Its efficient array operations, along with its seamless integration with other libraries, make it a powerful and indispensable library for numerical computations in Python.

### np.array() vs   np.asarray()
- The main difference between np.asarray() and np.array() is how they handle the input data type and whether they create a new array or return a reference to the input data. Here's a brief explanation:

#### np.array():

np.array() creates a new NumPy array from the input data.
It always creates a new copy of the input data, even if the input is already a NumPy array.
If the input is a nested sequence (e.g., a list of lists), np.array() will create a multidimensional array.
If the input is not a sequence, np.array() will try to interpret it as a scalar and create a 0-dimensional array.
np.array() is more commonly used when you want to create a new array from scratch or ensure that a new copy is created.


#### np.asarray():

np.asarray() is used to convert the input to a NumPy array, but it does not always create a new copy.
If the input is already a NumPy array, np.asarray() will return a reference to the input array without creating a new copy.
If the input is not a NumPy array, np.asarray() will create a new array object.
np.asarray() is useful when you want to convert an input to a NumPy array without unnecessarily creating a new copy, especially if the input might already be a NumPy array.


```python
import numpy as np

# Input list
my_list = [1, 2, 3, 4, 5]

# Using np.array()
arr1 = np.array(my_list)
arr2 = np.array(arr1)

# Using np.asarray()
arr3 = np.asarray(my_list)
arr4 = np.asarray(arr1)

print(arr1 is arr2)  # False
print(arr1 is arr3)  # False
print(arr1 is arr4)  # True
```

### numpy.argsort()

- Returns the indices that would sort an array.
- Perform an indirect sort along the given axis using the algorithm specified by the kind keyword. It returns an array of indices of the same shape as a that index data along the given axis in sorted order.

```python
# input array
in_arr = np.array([ 2, 0,  1, 5, 4, 1, 9])
print ("Input unsorted array : ", in_arr)

out_arr = np.argsort(in_arr)
print ("Output sorted array indices : ", out_arr)
print("Output sorted array : ", in_arr[out_arr])

## Indices lowest two values
np.argsort(in_arr)[:2]
```

```python
## np.argsort in decending order
res=np.argsort(in_arr)[::-1]
print(res)
```

## write a python function two calculate Euclidean distance between two points

#### In NumPy, you can extend or shrink the dimensions of an array using various functions and methods. Here are a few common ways to extend or shrink the axes of a NumPy array:

np.expand_dims() - Extend Axis:
The np.expand_dims() function is used to add a new axis to an array at a specified position. It takes the array and the axis position as arguments. Here's an example:

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr.shape)  # Output: (5,)

expanded_arr = np.expand_dims(arr, axis=0)
print(expanded_arr.shape)  # Output: (1, 5)
```

#### np.squeeze() - Shrink Axis:
The np.squeeze() function is used to remove single-dimensional axes from an array. It returns a new array with the same data but with the single-dimensional axes removed. Here's an example:

```python
import numpy as np

arr = np.array([[1, 2, 3, 4, 5]])
print(arr.shape)  # Output: (1, 5)

squeezed_arr = np.squeeze(arr)
print(squeezed_arr.shape)  # Output: (5,)
```
