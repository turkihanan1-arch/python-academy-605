In Multiprocessing, instead of utilizing a single core of the processor, utilizing all or more core to solve the problem. As a result, this reduces the time computation. Many different algorithms have been designed for multiplying matrices in parallel fashion. For example, block matrix multiplcation works by dividing the matrices into blocks and each thread calculates results of one block only.
