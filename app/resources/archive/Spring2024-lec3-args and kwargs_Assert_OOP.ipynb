# Part 1: Python args and kwargs

In Python, *args and **kwargs are special syntaxes used in function definitions. They allow a function to accept a variable number of arguments, with *args used for positional arguments and **kwargs for keyword arguments.

### Suppose, we define a function for addition of 3 numbers.

```python
def adder(x,y,z):
    print("sum:",x+y+z)

adder(10,12,13)
```

### Lets see what happens when we pass more than 3 arguments in the adder() function

```python
def adder(x,y,z):
    print("sum:",x+y+z)

adder(5,10,15,20,25)
```

* <font color='blue'><h3>__*args__ allows you to pass a varying number of positional arguments. Take the following example:

```python
# sum_integers_args.py
def my_sum(*args):
    result = 0
    # Iterating over the Python args tuple
    for x in args:
        result += x
    return result

print(my_sum(1, 2, 3))
print(my_sum(5,10,15,20,25))

```

```python
def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3))  # Output: 6
print(add_numbers(1, 2, 3, 4, 5))  # Output: 15
```

# The **kwargs

The **kwargs allows you to pass a variable number of keyword arguments to a function. kwargs stands for keyword arguments, and it is usually used to handle named arguments in a function. Inside the function, kwargs is a dictionary where the keys are the argument names, and the values are the argument values.

#### When a function has the **kwargs parameter, it can accept a variable number of keyword arguments as a dictionary.

```python
def concatenate(**kwargs):
    result = ""
    # Iterating over the Python kwargs dictionary
    for v in kwargs.values():
        result += v
    return result

print(concatenate(a="Real", b="Python", c="Is", d="Great", e="!"))
```

```python
def greet(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")

greet(first_name="John", last_name="Doe")  
# Output:
# first_name : John
# last_name : Doe
```

# Part 2: Python Assert Statement

The Python assert statement is a debugging aid that tests a condition as an internal self-check in your program. The statement is used to confirm that a particular condition is True. If the condition is False, the program will raise an AssertionError exception with the possibility to provide a message.

## Syntax for using Assert in Pyhton:

```python
def divide(dividend, divisor):
    assert divisor != 0, "Divisor cannot be zero."
    return dividend / divisor

# Correct usage (divisor is not zero)
result = divide(10, 2)
print(result)  # Output: 5.0
```

```python

# This will raise an assertion error because the divisor is zero
result = divide(10, 0)
```

-  Object-oriented programming combines a group of variables (properties) and functions (methods) into a unit called an object. These objects are organized into classes where individual objects can be grouped together. ### Class Methods

### In Object-oriented programming, Inside a Class, we can define the following three types of methods.

* Instance method: Used to access or modify the object state. If we use instance variables inside a method, such methods are called instance methods.
* Class method: Used to access or modify the class state. In method implementation, if we use only class variables, then such type of methods we should declare as a class method.

```python
# class methods demo
class Student:
    # class variable
    school_name = 'ABC School'

    # constructor
    def __init__(self, name, age):
        # instance variables
        self.name = name
        self.age = age

    # instance method
    def show(self):
        # access instance variables and class variables
        print('Student:', self.name, self.age, Student.school_name)

    # instance method
    def change_age(self, new_age):
        # modify instance variable
        self.age = new_age

    # class method
    @classmethod
    def modify_school_name(cls, new_name):
        # modify class variable
        cls.school_name = new_name

s1 = Student("Harry", 12)

# call instance methods
s1.show()
s1.change_age(14)

# call class method
Student.modify_school_name('XYZ School')
# call instance methods
3 s1.show()
```

```python
# Parent class
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def display_info(self):
        print(f"This is a {self.brand} vehicle from {self.year}.")

# Child class
class Car(Vehicle):
    def __init__(self, brand, year, model):
        super().__init__(brand, year)  # Call the __init__ of the parent Vehicle class
        self.model = model  # Additional property for Car

    def display_car_info(self):
        print(f"This is a {self.brand} {self.model} from {self.year}.")

# Create an instance of Car
my_car = Car("Toyota", 2020, "Corolla")

# Call method from the parent class
my_car.display_info()        # Output: This is a Toyota vehicle from 2020.

# Call method from the child class
my_car.display_car_info()    # Output: This is a Toyota Corolla from 2020.
```

# __call__ and __getitem__ methods


The ____call____ method can be defined in any class to make instances of that class callable. It can take arguments just like any function. Here is a simple example:

```python
class Adder:
    def __init__(self, n):
        self.n = n

    def __call__(self, x):
        return self.n + x

# Create an instance of Adder
add_five = Adder(5)

# Use the instance as if it were a function
print(add_five(10))  # Output: 15
```

# Example of __getitem__
The __getitem__ method is used to access items using the subscription syntax. For example, if you want to create a class that mimics the behavior of a list or a dictionary, __getitem__ can be helpful. Here's an example for a class that represents a simple infinite arithmetic sequence:

```python
class MyList:
    def __init__(self, data):
        self.data = data

    def __getitem__(self, index):
        return self.data[index]

# Creating an instance of the class
my_list = MyList([1, 2, 3, 4, 5])

# Accessing elements using indexing
element = my_list[2]
print(element)  # Output: 3
```

```python
class ReverseRange:
    def __init__(self, start, end):
        self.current = end
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.start:
            raise StopIteration
        self.current -= 1
        return self.current + 1

# Using the ReverseRange class
reverse_range = ReverseRange(1, 5)

for num in reverse_range:
    print(num)

# This will output:
# 5
# 4
# 3
# 2
# 1
```
