# Pandas

##### Pandas is a powerful, open-source Python library that has become an essential tool for data manipulation and analysis. Developed by Wes McKinney in 2008, Pandas stands for "Python Data Analysis Library" and is built on top of the NumPy library. It provides highly optimized performance with back-end source code written in C and Cython.

### Key Features of Pandas
- DataFrame Object: At the heart of Pandas is the DataFrame, a two-dimensional data structure similar to a spreadsheet or SQL table. It is designed to handle a diverse range of data types and supports both labeled rows and columns, making data manipulation intuitive and accessible.

- Efficient Data Manipulation: Pandas provides numerous functions for efficiently performing operations like merging, reshaping, selecting, as well as slicing of large data sets. It helps in handling both time series and non-time series data.

- Handling Missing Data: Pandas simplifies the process of detecting, removing, replacing, and manipulating missing data. It is well-equipped to handle NaN and None values, which are often a headache in data processing.

- Time Series Functionality: Time series tools are another strong suit of Pandas, making it a great choice for applications in finance, statistics, economics, and other areas where time series data is crucial.

- Read and Write Support: Pandas supports a wide range of file formats including CSV, Excel, JSON, HTML, and HDF5, as well as SQL databases, making data loading and saving both straightforward and versatile.

- Efficiently Joining and Merging Data: The library provides multiple ways to combine and join data from different sources, which is essential in many data analysis workflows.

In pandas, both pd.loc and pd.iloc are indexing methods used to access data within a DataFrame. However, they differ in how they reference and select data based on different indexing approaches.

### pd.loc:

- pd.loc allows indexing and selection of data based on label-based indexing. It uses labels or names of rows and columns to access data. The syntax for pd.loc is df.loc[row_label, column_label].
- The labels can be single values, lists, slices, or boolean arrays. For example:
    - df.loc[3, 'column_name']: Access the value at row 3 and 'column_name'.
    - df.loc[:, 'column_name']: Access all rows for 'column_name'.
    - df.loc[1:5, ['column1', 'column2']]: Access rows 1 to 5 for columns 'column1' and 'column2'.
pd.iloc:

### pd.iloc
- pd.iloc allows indexing and selection of data based on integer-based indexing. It uses integer positions of rows and columns to access data. The syntax for pd.iloc is df.iloc[row_index, column_index].
- The indices can be single integers, lists of integers, slices, or boolean arrays. For example:
    - df.iloc[0, 2]: Access the value at the first row and third column.
    - df.iloc[:, 1]: Access all rows for the second column.
    - df.iloc[1:5, [0, 2]]: Access rows 1 to 4 for columns at positions 0 and 2.


###### To summarize, pd.loc uses label-based indexing, allowing you to access data using row and column labels/names, while pd.iloc uses integer-based indexing, allowing you to access data using row and column positions/indices.

### Introduction to Filtering in Pandas

Filtering data in Pandas is a fundamental task when working with data frames in Python. You use filtering to select specific rows from a DataFrame based on the values of one or more columns. This process can help you analyze subsets of data, detect patterns, remove or handle outliers, or prepare data for further processing like visualization or statistical analysis.

Here’s a step-by-step guide on how to filter data using Pandas, including some common scenarios.



Then, you can import Pandas and create a sample DataFrame:


```python

import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles']
}
df = pd.DataFrame(data)
df
```

```python


### Filtering Techniques

#### 1. **Basic Filtering**

#To filter rows where a column meets a specific condition:


# Filter rows where Age is greater than 30
filtered_df = df[df['Age'] > 30]
print(filtered_df)
```

```python


#### 2. **Multiple Conditions**

#You can combine more than one condition using logical operators like `&` (and), `|` (or):


# Filter rows where Age is greater than 30 and City is New York
filtered_df = df[(df['Age'] > 30) & (df['City'] == 'New York')]
print(filtered_df)
```

```python


#### 3. **Using `query()` Method**

### Pandas also provides the `query()` method which can be more readable, especially with complex conditions:


# Using query to filter data
filtered_df = df.query('Age > 30 & City == "New York"')
print(filtered_df)

```

```python

#### 4. **Filtering Based on String Conditions**

##If you need to filter by string operations, you can use the `str` accessor:


# Filter rows where the Name starts with 'A'
filtered_df = df[df['Name'].str.startswith('A')]
print(filtered_df)
```

```python


#### 5. **Filtering with `isin()`**

##If you need to check if values are within a specific set, use `isin()`:


# Filter rows where City is either New York or Chicago
filtered_df = df[df['City'].isin(['New York', 'Chicago'])]
print(filtered_df)
```

```python


#### 6. **Filtering Using `isnull()` or `notnull()`**

#These are useful for finding rows where data may be missing or where data is present:


# Assuming there are missing values in 'City'
df['City'] = df['City'].replace('Chicago', None)

# Filter rows where City is not null
filtered_df = df[df['City'].notnull()]
print(filtered_df)

# Filter rows where City is null
null_cities_df = df[df['City'].isnull()]
print(null_cities_df)

```

```python
import pandas as pd

df = pd.DataFrame({
    'name':['john','mary','peter','jeff','bill','lisa','jose'],
    'age':[23,78,22,19,45,33,20],
    'gender':['M','F','M','M','M','F','M'],
    'state':['california','dc','california','dc','california','texas','texas'],
    'num_children':[2,0,0,3,2,1,4],
    'num_pets':[5,1,0,5,2,2,3]
})
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# a scatter plot comparing num_children and num_pets
df.plot(kind='scatter',x='num_children',y='num_pets',color='red')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# a simple line plot
df.plot(kind='bar',x='name',y='age')
```

# For machine learning train-test split

```python
link = 'http://bit.ly/uforeports'
ufo = pd.read_csv(link)
```

```python
train = ufo.sample(frac=0.75, random_state=99)
```

```python
# you can't simply split 0.75 and 0.25 without overlapping
# this code tries to find that train = 75% and test = 25%
test = ufo.loc[~ufo.index.isin(train.index), :]
```
