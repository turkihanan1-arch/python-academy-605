# 1-  Introduction to Json

### What is JSON?

- JSON (JavaScript Object Notation) is a lightweight data format used for data interchange.
- It is easy for humans to read and write, and easy for machines to parse and generate.
 - JSON is language-independent but uses conventions familiar to programmers of the C-family of languages, including C, C++, C#, Java, JavaScript, Perl, Python, and others.

#### Overview of Python's json Module

######  Python’s built-in json module can be used to encode and decode JSON data, also known as serialize and deserialize.
#### Key functions:
 - json.loads(): Deserialize a JSON document from a string.
- json.dumps(): Serialize Python objects into a JSON formatted string.
- json.load(): Deserialize JSON from a file.
- json.dump(): Serialize Python objects to a file using JSON.

# Reading json data
### Deserializing JSON (Reading JSON)

##### - json.loads() example: Convert JSON string to Python object.




```python
import json

# Example JSON string
json_string = '{"name": "John", "age": 30, "city": "New York"}'
data = json.loads(json_string)

print(data)
```

#### json.load() example: Load JSON data from a file.

```python



import json

# Assuming `data.json` contains `{"name": "Jane", "age": 25, "city": "Los Angeles"}`
with open('airports.json', 'r') as file:
    data = json.load(file)

#print(data)
```

## Writing JSON Data
#### Serializing JSON (Writing JSON)

###### json.dumps() example: Convert Python object to JSON string.

```python
import json

# Example JSON string
json_string = '{"name": "John", "age": 30, "city": "New York"}'
data = json.loads(json_string)

print(data)
```

```python
import json

data = {'name': 'Bob', 'age': 29, 'city': 'Miami'}
with open('output.json', 'w') as file:
    json.dump(data, file)
```

### Beautifying JSON Output

#### Use indent and sort_keys in json.dumps() and json.dump() for better formatting.

```python
import json

data = {'name': 'Carol', 'age': 24, 'city': 'Seattle'}
formatted_string = json.dumps(data)

print(formatted_string)
```

```python
import json

data = {'name': 'Carol', 'age': 24, 'city': 'Seattle'}
formatted_string = json.dumps(data, indent=4, sort_keys=True)

print(formatted_string)
```

## What is Git?
- Git is a tool that helps you manage and track changes in your code projects.
- It allows multiple people to work on the same project without overwriting each other's work.
- Git keeps a history of all the changes made to your code, so you can go back to previous versions if needed.

## Basic Git Commands
- `git init`: Start a new Git project.
- `git add`: Tell Git to track changes in specific files.
- `git commit`: Save a snapshot of your code changes with a message describing what you did.
- `git push`: Send your code changes to a remote location (like GitHub).
- `git pull`: Get the latest changes from the remote location.

## What is GitHub?
- GitHub is a website where you can store your Git projects online.
- It makes it easy to collaborate with others on the same project.
- GitHub also provides extra features like bug tracking, project management, and a place to host websites.

3. Make changes to your code on your computer.
4. Use `git add` and `git commit` to save your changes.
5. Use `git push` to send your changes back to GitHub.
6. Other people can then see your changes and contribute to the project.

## Why Use Git and GitHub?
- They help you keep your code organized and backed up.

By using Git and GitHub, you can become a more efficient and collaborative programmer, even if you're just starting out!

# 3- Sending email

In Python, sending emails can be efficiently handled using the smtplib library along with the EmailMessage class from the email.message module. This approach allows you to easily construct and send emails from your Python scripts, offering a straightforward way to automate email communications. Whether you're sending notifications, alerts, or regular updates, these tools provide a robust foundation for email handling in Python.



```python
import smtplib
from email.message import EmailMessage
```

```python


email_sender = 'sender@example.com'
email_password = 'sender_password'
email_receiver = 'receiver@example.com'

subject = 'Email subject'
body = 'Email body'

em = EmailMessage()
em['From'] = email_sender
em['To'] = email_receiver
em['Subject'] = subject
em.set_content(body)

with smtplib.SMTP_SSL('smtp.example.com', 465) as smtp:
  smtp.login(email_sender, email_password)  
  smtp.send_message(em)
```

#### To break this down:

- We import smtplib for sending mail, and EmailMessage from email to construct the message.
- Specify the sender, receiver, email subject and body content.
- Create an EmailMessage object, set the From, To, Subject fields and email content.
- Connect to the SMTP server using smtplib and login using the email sender and password.
- Send the email message using smtp.send_message().
- The with statement takes care of closing the connection after sending the email.
- Make sure to use your actual sender, receiver email IDs, SMTP server details and login credentials.

## Introduction to Reading PDF Files Using PyPDF2
- Working with PDF files programmatically can be essential for various tasks, such as extracting text for data analysis, automating report generation, or even modifying PDFs to fit specific requirements. Python offers several libraries for handling PDF files, and one of the most popular and accessible is PyPDF2. This library provides a wide range of functionalities, including reading text from PDFs, merging and splitting PDF files, and more.

### PyPDF2


- pypdf2 is a Pure-Python library built as a PDF toolkit. It is capable of extracting document information, splitting documents page by page, merging documents page by page, cropping pages, merging multiple pages into a single page, encrypting and decrypting PDF files, and more. In this guide, we will focus on how to use PyPDF2 to read PDF files, specifically extracting text from them.



```python
!pip install PyPDF2
```

```python
import PyPDF2

def read_all_pages(pdf_path):
    with open(pdf_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        num_pages = len(reader.pages)
        
        for page_num in range(num_pages):
            page = reader.pages[page_num]
            text = page.extract_text()
            print(f"Page {page_num + 1}:")
            print(text)
            print("------------------------")

def read_specific_page(pdf_path, page_num):
    with open(pdf_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        
        if 1 <= page_num <= len(reader.pages):
            page = reader.pages[page_num - 1]
            text = page.extract_text()
            print(f"Page {page_num}:")
            print(text)
        else:
            print(f"Invalid page number. The PDF has {len(reader.pages)} pages.")
# Example usage:
pdf_path = 'page.pdf'
print("All text from PDF:")
read_all_pages(pdf_path)

read_specific_page(pdf_path, 1)
```

### Stacking subplots in two directions
When stacking in two directions, the returned axs is a 2D numpy array.

If you have to set parameters for each subplot it's handy to iterate over all subplots in a 2D grid using for ax in axs.flat:.



```python

# Some example data to display
x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x ** 2)
```

```python
fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x, y)
axs[0, 0].set_title('Axis [0,0]')
axs[0, 1].plot(x, y, 'tab:orange')
axs[0, 1].set_title('Axis [0,1]')
axs[1, 0].plot(x, -y, 'tab:green')
axs[1, 0].set_title('Axis [1,0]')
axs[1, 1].plot(x, -y, 'tab:red')
axs[1, 1].set_title('Axis [1,1]')


for ax in axs.flat:
    ax.set(xlabel='x-label', ylabel='y-label')

# Hide x labels and tick labels for top plots and y ticks for right plots.
for ax in axs.flat:
    ax.label_outer()
    
plt.show()
```

### Sharing axes
By default, each Axes is scaled individually. Thus, if the ranges are different the tick values of the subplots do not align.

```python
fig, (ax1, ax2) = plt.subplots(2)
fig.suptitle('Axes values are scaled individually by default')
ax1.plot(x, y)
ax2.plot(x + 1, -y)
plt.show()
```

# Plotting a 3D Scatter Plot in Matplotlib
If you don't want to visualize this in two separate subplots, you can plot the correlation between these variables in 3D. Matplotlib has built-in 3D plotting functionality, so doing this is a breeze.

First, we'll need to import the Axes3D class from mpl_toolkits.mplot3d. This special type of Axes is needed for 3D visualizations. With it, we can pass in another argument - z, which is the third feature we'd like to visualize.



```python
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D

df = pd.read_csv('AmesHousing.csv')

fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')

x = df['SalePrice']
y = df['Gr Liv Area']
z = df['Overall Qual']

ax.scatter(x, y, z)
ax.set_xlabel("Sale price")
ax.set_ylabel("Living area above ground level")
ax.set_zlabel("Overall quality")

plt.show()
```
