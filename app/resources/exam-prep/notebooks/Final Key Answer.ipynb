# Q3: (7 Marks)

## Part 1 (2 Marks)
## Create and print a DataFrame with the following data:

| Name | Score |
| --- | --- |
| Ahmad | 75 |
| Ali | 80 |
| Sara | 95 |
| Omar | 70 |



````python
# write your code here

data = {
    "Name": ["Ahmad", "Ali", "Sara", "Omar"],
    "Score": [75, 80, 95, 70]
}

df = pd.DataFrame(data)
print(df)
````

# Part 2 (3 Mark)

## Add a new column "Result":
* "Pass" if Score ≥ 75
* "Fail" otherwise

````python
# write your code here


df["Result"] = ( (df["Score"] >= 75) )
df["Result"] = df["Result"].replace({True: "Pass", False: "Fail"})
print(df)


############

df["Result"] = df["Score"].apply(lambda x: "Pass" if x >= 75 else "Fail")
print(df)


############

df["Result"] = ["Pass" if x >= 75 else "Fail" for x in df["Score"]]
print(df)

#############

# Add Result column using numpy where
df["Result"] = np.where(df["Score"] >= 75, "Pass", "Fail")
````

# Part 3 (2 Marks)

## Plot a bar graph for scores
Add:

````python
# write your code here


# Plot bar graph for scores

plt.bar(df["Name"], df["Score"])

plt.title("Student Scores")
plt.xlabel("Name")
plt.ylabel("Score")

plt.grid(True)
plt.show()
````

## Given the following NumPy array:




Write Python code to perform the following operations:

* Display all even numbers in the array. (0.5 mark)
* Find and display the maximum value. (0.5 mark)
* Find and display the minimum value. (0.5 mark)
* Calculate and display the average (mean) of the array. (0.5 mark)

````python
# Given:
arr = np.array([1, 3, -5, 7, 9 , -3 , 0 , 4, 10])

# write your code here

# even numbers    #(0.5 Mark)
even_numbers = arr[arr % 2 == 0]
print("Even numbers:", even_numbers)

# max, min, average
print("Maximum:", np.max(arr))    #(0.5 Mark)
print("Minimum:", np.min(arr))    #(0.5 Mark)
print("Average:", np.mean(arr))   #(0.5 Mark)
````

## Write a Python script using NumPy, Pandas, and Matplotlib to perform the following tasks:


## Part 1: Data Generation (2.5 marks)

Create a DataFrame with 50 rows containing the following columns:

* "ID" → sequential values from 2602001 to 2602050 (0.5 mark)
* "Math" → 50 random integers between 25 and 99 (0.5 mark)
* "Physics" → 50 random integers between 25 and 99 (0.5 mark)
* "Chemistry" → 50 random integers between 25 and 99 (0.5 mark)

Display Last 8 Rows    (0.5 mark)


````python
# write your code here

df = pd.DataFrame({
    "ID": np.arange(2602001, 2602051),
    "Math": np.random.randint(25, 100, 50),
    "Physics": np.random.randint(25, 100, 50),
    "Chemistry": np.random.randint(25, 100, 50)
})

print(df.tail(8))
````

## Part 2: Data Processing (2 marks)
After add a new column "Average" representing the mean of the three subjects

* Display the top 5 students with the highest average (1 mark)
* Display the number of students whose average is below the overall average column (1 mark)


````python
# Given
# Average column
df["Average"] = df[["Math", "Physics", "Chemistry"]].mean(axis=1)

# write your code here

# (1 Mark)
# Top 5 students
print("Top 5 Students:")
print(df.sort_values(by="Average", ascending=False).head(5))


# (1 Mark)
# below overall average
overall_avg = df["Average"].mean()

count_below = (df["Average"] < overall_avg).sum()
print("Below overall average: ",count_below)

````

## Part 3: Visualization (3.5 marks)

Using the same DataFrame:

1. Create one figure of size (7, 4) that contains two plots (0.5 mark)

2. First plot:
  * A pi chart showing the average score of each subject (Math, Physics, Chemistry) (1 mark)

3. Second plot:
  * A boxplot for the columns: Math, Physics, Chemistry (1 mark)

4. Add the following to the all charts:
  * Title (0.25 mark)
  * Axis labels (0.25 mark)
  * Grid (0.25 mark)
  * Apply gray style (0.25 mark)

````python
# write your code here

plt.style.use("ggplot")

# Create figure
plt.figure(figsize=(7, 4))

# -------------------
# 1st Plot (Pie Chart)
# -------------------
plt.subplot(1, 2, 1)

subject_means = df[["Math", "Physics", "Chemistry"]].mean()

plt.pie(subject_means.values,
        labels=subject_means.index,
        autopct="%1.1f%%",
        colors=["red", "blue", "green"])

plt.title("Average Score per Subject")

# -------------------
# 2nd Plot (Boxplot)
# -------------------
plt.subplot(1, 2, 2)

plt.boxplot([df["Math"], df["Physics"], df["Chemistry"]],
            labels=["Math", "Physics", "Chemistry"])

plt.title("Score Distribution")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(True)

# Final layout
plt.tight_layout()
#plt.show()
````

## Given the following list of numbers:

numbers = {

    "ID": ["A", "B", "C", "D", "E", "F", "G"],

}

## Tasks:
* Write the numbers into a CSV file named "data.csv" using the pandas library. (1 mark)
* Read the data back from the file, compute their sum, and print the result. (2 mark)


````python
# Given
numbers = {
    "ID": ["A", "B", "C", "D", "E", "F", "G"],
    "$": [10, 20, 30, 40, 50, 60, 70]
}

# write your code here


# Create DataFrame
df = pd.DataFrame(numbers)

# Write data to CSV file
df.to_csv("data.csv", index=False)

# Read data from CSV file
data = pd.read_csv("data.csv")

# Compute the sum of the $ column
total = data["$"].sum()

# Print the result
print("Sum =", total)
````

# Q7 (5 marks):

## Given the following students’ scores:

|Subject | CS | Math | Physics |
| --- | --- |--- | --- |
| S1 | 77 | 80 | 72 |
| S2 | 80 | 80 | 85 |
| S3 | 90 | 90 | 83 |
| S4 | 70 | 70 | 72 |




## Tasks:
* Write the table into a 2D Numpy Array. (1 mark)
* Increase all Physics marks by 5. (1 mark)
* Print the sub-table formed by rows S2 and S3 and columns CS and Math. (1 mark)
* Print the unique values in the matrix without associated indices. (1 mark)
* Print Shape of the Array, Number of Dimensions of the Array, Total Number of Elements in the Array & Data Type of Array Elements. (1 mark)

````python
# write your code here


# Create 2D NumPy array
scores = np.array([
    [77, 80, 72],
    [80, 80, 85],
    [90, 90, 83],
    [70, 70, 72]
])

# Increase all Physics marks by 5
scores[:, 2] = scores[:, 2] + 5

# Print sub-table (rows S2 & S3, columns CS & Math)
print("\nSub-table:")
print(scores[1:3, 0:2])

# Print unique values
print("\nUnique Values:")
print(np.unique(scores))

# Print array properties
print("\nShape of the Array:", scores.shape)
print("Number of Dimensions:", scores.ndim)
print("Total Number of Elements:", scores.size)
print("Data Type of Elements:", scores.dtype)
````
