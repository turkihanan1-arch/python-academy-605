### 4- Write a function that takes a list as input as following:  (5 marks)
Numbers = [11, 2, 44, 54, 5, 16, 17, 8, 19].

### Perform four specific operations:

- Display even numbers
- Display sum of odd numbers
- Display the maximum
- Display the minimum
- Display the average

````python
# Define the function here
def analyze_numbers(numbers):
    # Even numbers
    even_numbers = [x for x in numbers if x % 2 == 0]
    print("Even numbers:", even_numbers)

    # Sum of odd numbers
    odd_sum = sum(x for x in numbers if x % 2 != 0)
    print("Sum of odd numbers:", odd_sum)

    # Maximum number
    print("Maximum number:", max(numbers))

    # Minimum number
    print("Minimum number:", min(numbers))

    print("Average:", sum(numbers) / len(numbers))

# Given list
Numbers = [11, 2, 44, 54, 5, 16, 17, 8, 19]

# Call the function
analyze_numbers(Numbers)
````

### 8- Write a Python program that defines a class called Weather. (5 marks)

Inside the class, define a method check_weather(self, temp) that takes a temperature value in Celsius as input and prints a message describing the weather according to the following conditions: (3 marks)

- If the temperature is less than or equal to 15, print:
"The weather is cold."
- else if the temperature is greater than 15 and less than 25, print:
"The weather is moderate."
- else if the temperature is greater than or equal to 25 and less than 35, print:
"The weather is fine."
- else if the temperature is greater than or equal to 35, print:
"The weather is hot."

Create an object of the class, read the temperature from the user, and call the method using the object. (2 marks)

````python
# Write your code here
class Weather:
    # Regular instance method
    def check_weather_instance(self, temp):
        if temp <= 15:
            print("Instance method: The weather is cold.")
        elif temp < 25:
            print("Instance method: The weather is moderate.")
        elif temp < 35:
            print("Instance method: The weather is fine.")
        else:
            print("Instance method: The weather is hot.")

weather_obj = Weather()
temperature = float(input("Enter temperature in Celsius: "))
weather_obj.check_weather_instance(temperature)
````

### 9-  Vehicle Rental System ( 13 marks)
### You are required to design a simple vehicle rental system using object-oriented programming with inheritance in Python.

### Part 1: Base Class (Vehicle) – (3 Marks)

Create a class named Vehicle that includes:

- Attributes:
  * vehicle_id (integer)
  * brand (string)
  * rental_price_per_day (float)
  * is_available (boolean, default = True)

- Methods:
  * A constructor to initialize all attributes. (1 mark)
  * A method rent_vehicle() that changes availability to False if the vehicle is available  and print its case. (1 mark)
  * A method display_info() to print vehicle details. (1 mark)

### Part 2: Derived Class (Car) – (4 Marks)

Create a class Car that inherits from Vehicle and adds: (2 mark)

- Attribute:
  * number_of_doors (integer)

- Method:
  * Override display_info() to include the number of doors. (2 marks)

### Part 3: Rental System Class – (3 Marks)

Create a class RentalSystem that manages one vehicle only:

- Attribute:
  * vehicle: stores a single Vehicle object
- Methods:
  * A constructor that receives a Vehicle object and stores it. (1 mark)
  * rent_vehicle(): calls the vehicle’s rent method. (1 mark)
  * show_vehicle(): displays the vehicle information. (1 mark)


  ### Part 4: Test the system (3 marks)
  - Create a car object
  - Create rental system with one vehicle
  - Display vehicle info


````python
# write your code here
# Base Class
class Vehicle:
    def __init__(self, vehicle_id, brand, rental_price_per_day, is_available=True):
        self.vehicle_id = vehicle_id
        self.brand = brand
        self.rental_price_per_day = rental_price_per_day
        self.is_available = is_available

    def rent_vehicle(self):
        if self.is_available:
            self.is_available = False
            print("Vehicle has been rented.")
        else:
            print("Vehicle is already rented.")

    def display_info(self):
        print(f"ID: {self.vehicle_id}, Brand: {self.brand}, Price/day: {self.rental_price_per_day}, Available: {self.is_available}")


# Derived Class
class Car(Vehicle):
    def __init__(self, vehicle_id, brand, rental_price_per_day, number_of_doors, is_available=True):
        super().__init__(vehicle_id, brand, rental_price_per_day, is_available)
        self.number_of_doors = number_of_doors

    def display_info(self):
        super().display_info()
        print(f"Doors: {self.number_of_doors}")


# Simplified Rental System (only one vehicle)
class RentalSystem:
    def __init__(self, vehicle):
        self.vehicle = vehicle

    def rent_vehicle(self):
        self.vehicle.rent_vehicle()

    def show_vehicle(self):
        self.vehicle.display_info()



# ------------------ Testing ------------------

# Create a car
car1 = Car(1, "Toyota", 100.0, 4)

# Create rental system with one vehicle
system = RentalSystem(car1)

# Display vehicle
system.show_vehicle()

# Rent vehicle
system.rent_vehicle()

# Display again
system.show_vehicle()

````
