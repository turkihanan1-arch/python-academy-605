### 9- Student Scores Analysis System. (13 marks)

Design a simple system using Python classes, inheritance, and NumPy.

The system manages **one dataset**. Its score array has one row per student and one column per subject. Assume that it is a nonempty, two-dimensional numeric array with scores between 0 and 100.

#### Part 1: Base class `Dataset` (3 marks)

Create a class named `Dataset` with:

- A constructor receiving `dataset_name` (string) and `scores` (NumPy array), and storing both as instance attributes. (1 mark)
- A method `display_info()` that prints the dataset name and the array shape. (1 mark)
- A method `show_scores()` that prints the score array. (1 mark)

#### Part 2: Derived class `StudentScores` (4 marks)

Create a class named `StudentScores` that inherits from `Dataset`:

- Add an attribute `subject_names`, a list containing one name per score column. Initialize the inherited attributes using `super()`. (2 marks)
- Override `display_info()` to print the dataset name, array shape, and subject names. (2 marks)

#### Part 3: NumPy analysis methods (3 marks)

Add the following methods to `StudentScores`. Use NumPy array operations without loops:

- `student_averages()`: return the mean score of each student. (1 mark)
- `subject_averages()`: return the mean score of each subject. (1 mark)
- `highest_score_location()`: return the position of the highest score as a zero-based `(row, column)` tuple. If tied, return the first position in row-major order. (1 mark)

#### Part 4: Test the system (3 marks)

Use the score data supplied in the next cell:

- Create a `StudentScores` object named `report`, with dataset name `Practice Scores` and subject names `['Python', 'NumPy', 'Statistics']`. (1 mark)
- Call `display_info()` and `show_scores()`. (1 mark: 0.5 each)
- Call the three analysis methods and print their returned results with clear labels. (1 mark: 1/3 per method)


````python
# Define Dataset and StudentScores here

# ---------- Part 1: base class ----------
class Dataset:
    # Your mistake: you named it dataName. Copy required names exactly: dataset_name
    def __init__(self, dataset_name, scores):
        self.dataset_name = dataset_name
        self.scores = scores

    def display_info(self):
        print("Dataset:", self.dataset_name)
        # shape = (rows, columns). Your mistake: you printed the whole array here.
        print("Shape:", self.scores.shape)

    def show_scores(self):
        print("Scores:")
        print(self.scores)


# ---------- Part 2: derived class ----------
class StudentScores(Dataset):
    def __init__(self, dataset_name, scores, subject_names):
        # super() lets the parent stick its two attributes. No self inside super().
        super().__init__(dataset_name, scores)
        self.subject_names = subject_names

    # Override: same name as the parent's method.
    # Run the parent's version first, then add the subjects.
    def display_info(self):
        super().display_info()
        print("Subjects:", self.subject_names)

    # ---------- Part 3: NOT IN THE MIDTERM (NumPy) ----------
    def student_averages(self):
        return self.scores.mean(axis=1)

    def subject_averages(self):
        return self.scores.mean(axis=0)

    def highest_score_location(self):
        flat_index = np.argmax(self.scores)
        row, col = np.unravel_index(flat_index, self.scores.shape)
        return (int(row), int(col))
````

````python
scores = np.array([[72, 84, 91],
                   [65, 78, 88],
                   [93, 89, 76],
                   [58, 67, 82]])

# Create the object and test the system here

# ---------- Part 4 ----------
# Your mistake: you created the object from the PARENT (Dataset) with 2 values.
# Create it from the CHILD, with 3 values in the same order as its __init__:
#   name, then the scores array, then the subjects list.
report = StudentScores("Practice Scores", scores, ['Python', 'NumPy', 'Statistics'])

report.display_info()
report.show_scores()

# NumPy part (not in the midterm)
print("Student averages:", report.student_averages())
print("Subject averages:", report.subject_averages())
print("Highest score location:", report.highest_score_location())
````
