* Matplotlib Common options:
  - `Line Color`:
    * 'b' → blue
    * 'g' → green
    * 'r' → red
    * 'c' → cyan
    * 'm' → magenta
    * 'y' → yellow
    * 'k' → black
  - `Line Style`:
    * '-' → solid
    * '--' → dashed
    * '-.' → dash-dot
    * ':' → dotted
    * ' ' or 'None' → no line
  - `Marker`:
    * 'o' → circle
    * 's' → square
    * '^' → triangle up
    * 'v' → triangle down
    * '<' → triangle left
    * '>' → triangle right
    * 'x' → x mark
    * '+' → plus
    * '*' → star
    * 'd' → diamond
    * '.' → point
  * `Line Width`:
    * 2–3 → moderately thicker
    * 3–5 → bold, very visible (great for projectors)
