## 1. Figure Size

Control plot size using `figsize`.

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,5)
y = x**2

plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title('Custom Figure Size')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

## 2. Styles

Matplotlib provides built-in styles.



```
plt.style.use('default')        # normal style
plt.style.use('ggplot')         # ggplot style
plt.style.use('dark_background') # dark theme
plt.style.use('seaborn-v0_8')   # modern clean style
```



````python
plt.figure(figsize=(6,4))

plt.style.use('dark_background')

plt.plot(x,y , color='red', marker='o', linestyle=':')
plt.title('Style Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

````python
plt.figure(figsize=(6,4))

plt.style.use('ggplot')
plt.plot(x,y)
plt.title('Style Example')
plt.grid(True)
plt.show()
````

## 3. Multiple Lines

Plot multiple datasets on same graph.

````python
y2 = x*3

plt.plot(x,y,label='x^2',marker='s',linestyle=':',color = 'r')
plt.plot(x,y2,label='3x',marker='o',linestyle='--')
plt.plot(x,y2*3,label='3x*3',marker='*',linestyle='-.')

plt.legend()
plt.grid(True)
plt.title('Multiple Lines')
plt.show()
````

## 4. Histogram

Used to show data distribution.

````python
data = np.random.randn(1000)

plt.style.use('default')


plt.hist(data, bins=20 , color = 'blue')

plt.title('Histogram')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

## 5. Box Plot

Shows data spread (median, quartiles, outliers).

A box plot shows 5 key values:
1. Minimum (lowest value)
2. Q1 (25% point)
3. Median (50% point)
4. Q3 (75% point)
5. Maximum (highest value)

````python
plt.boxplot(data)

plt.title('Box Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

## 6. Saving Figures

Save plots as image files.

````python
plt.plot(x,y)
plt.title('Save Figure')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.savefig('my_plot.png')

plt.show()
````

Improves spacing between subplots.

# Matplotlib with Pandas

````python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500

df = pd.DataFrame({
    'ID': range(1, n + 1),
    'Math': np.random.randint(50, 100, n),
    'Physics': np.random.randint(40, 100, n),
    'Chemistry': np.random.randint(45, 100, n),
    'English': np.random.randint(50, 100, n),
    'Attendance': np.random.randint(60, 100, n)
})
# average
avg = df[["Math", "Physics", "Chemistry", "English"]].mean(axis=1)
# StudyHours depends on performance
df["StudyHours"] = ((avg / 10) + np.random.normal(0, 1, n)).clip(1, 10).astype(int)

# Save
df.to_csv('big_students.csv', index=False)

# Load again
df = pd.read_csv('big_students.csv')

print(df.head(10))
````

````python

avg_scores = df[['Math','Physics','Chemistry','English']].mean()

print(avg_scores)

plt.bar(avg_scores.index, avg_scores.values, color = 'red')
plt.title("Average Subject Scores")
plt.ylabel("Score")
plt.grid(True)

plt.show()
````

````python
df["Pass"] = (
    (df["Math"] >= 60) &
    (df["Physics"] >= 60) &
    (df["Chemistry"] >= 60) &
    (df["English"] >= 60)
)

print(df.head())

pass_count = df["Pass"].sum()
fail_count = len(df) - pass_count


plt.pie([pass_count, fail_count],
        labels=["Pass", "Fail"],
        autopct="%1.1f%%",
        startangle=90)

plt.title("Overall Pass vs Fail Students")
plt.show()
````

````python
plt.boxplot([
    df['Math'],
    df['Physics'],
    df['Chemistry'],
    df['English']
])

plt.title("Box Plot of Student Scores")
plt.xticks([1,2,3,4], ['Math','Physics','Chemistry','English'])
plt.xlabel("Courses")
plt.ylabel("Scores")
plt.grid(True , linestyle=":")
plt.show()
````

````python
plt.hist(df['Attendance'], color='red' , bins=20)
plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")
plt.grid(True , linestyle=":")

plt.show()
````

````python

# compute average
avg = df[["Math", "Physics", "Chemistry", "English"]].mean(axis=1)

# scatter plot
plt.scatter(df["StudyHours"], avg, alpha=0.4 , color = 'red')

plt.title("Study Hours vs Average Grade")
plt.xlabel("Study Hours")
plt.ylabel("Average Grade")

plt.grid(True, linestyle=":")
plt.show()
````
