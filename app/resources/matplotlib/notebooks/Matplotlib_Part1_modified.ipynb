# Matplotlib (Part 1)




## 1. What is Matplotlib?

- **Matplotlib** is a Python library used for **data visualization**.
- It helps us display data as **graphs and charts**.
- It is often used with:
  - **NumPy** for numerical data
  - **Pandas** for tables and datasets


```python
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Pyplot Basics (`matplotlib.pyplot`)

- `matplotlib.pyplot` is commonly imported as `plt`.
- It is used to create, control, and display plots.

### Simple example

```python
plt.plot([1, 2, 3], [4, 5, 6])

plt.show()
```

## 3. Line Plot

- A **line plot** is used to show **trends** or **continuous data**.
- It is one of the most important chart types in Matplotlib.

### Common options
- `color` → changes line color
  * 'b' → blue
  * 'g' → green
  * 'r' → red
  * 'c' → cyan
  * 'm' → magenta
  * 'y' → yellow
  * 'k' → black
- `linestyle` → changes line style
  * '-' → solid
  * '--' → dashed
  * '-.' → dash-dot
  * ':' → dotted
  * '' or 'None' → no line
- `marker` → adds symbols on points
  * 'o' → circle
  * 's' → square
  * '^' → triangle up
  * 'v' → triangle down
  * '<' → triangle left
  * '>' → triangle right
  * 'x' → x mark
  * '+' → plus
  * '*' → star
  * 'd' → diamond
  * '.' → point


```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 6, 8, 10]

plt.plot(x, y, color='green', marker='d', linestyle=':' )

plt.show()
```

## 4. Scatter Plot

- A **scatter plot** is used to show the **relationship between two variables**.
- Each point represents one data value.


```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 6, 8, 10]

plt.scatter(x, y)

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

colors = ['red', 'green', 'blue', 'orange']
sizes = [50, 200, 100, 300]

plt.scatter(x, y, c=colors, s=sizes)

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

colors = ['red', 'green', 'blue', 'orange']
sizes = [50, 200, 100, 300]

plt.scatter(x, y, c=colors, s=sizes)

plt.grid(True)
plt.show()
```

## 6. Grid

- `plt.grid(True)` adds grid lines.
- Grid lines make the chart easier to read.


```python
x = np.arange(0, 6)
y = x * 2

plt.plot(x, y, marker='o')

plt.grid(True)
plt.show()
```

## 5. Labels, Title, and Legend

These are very important because they make the graph clear and easy to understand.

- `plt.xlabel()` → label for x-axis
- `plt.ylabel()` → label for y-axis
- `plt.title()` → title of the graph
- `plt.legend()` → explains multiple lines or categories


```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]


plt.plot(x, y1, label='Linear', color='green', marker='o', linestyle=':')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Plot with Legend')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [1, 4, 9, 16, 25]
y3 = [10 , 20 , 15 , 20 , 10]

plt.plot(x, y1, label='Linear', color='green', marker='o', linestyle=':')
plt.plot(x, y2, label='Quadratic')
plt.plot(x, y3, label='xxx' , color='k', marker='o', linestyle='-.')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Plot with Legend')
plt.legend()
plt.grid(True)
plt.show()
```

## 7. Subplots

- **Subplots** allow us to place more than one graph in the same figure.
- This is useful for comparison.

plt.subplot(# rows, # column, current plot position)

Grid layout (2 rows × 2 columns):

(1) (2)

(3) (4)


```python
x = np.arange(1, 6)
y1 = x
y2 = x**2

plt.subplot(2, 1, 1)
plt.plot(x, y1, marker='o')
plt.title('First Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(x, y2, marker='s')
plt.title('Second Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.tight_layout()
plt.show()
```

```python
x = np.arange(1, 6)
y1 = x
y2 = x**2

plt.subplot(1, 2, 1)
plt.plot(x, y1, marker='o')
plt.title('First Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(x, y2, marker='s')
plt.title('Second Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.tight_layout()
plt.show()
```

```python
x = np.arange(1, 6)
y1 = x
y2 = x**2

plt.subplot(2, 2, 1)
plt.plot(x, y1, marker='o')
plt.title('First Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(x, y2, marker='s')
plt.title('Second Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.subplot(2, 2, 2)
plt.plot(x, y1, marker='o')
plt.title('3rd Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(x, y2, marker='s')
plt.title('4th Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.tight_layout()
plt.show()
```

## 8. Bar Chart

- A **bar chart** is used to compare **categories**.
- Each bar represents a value.


```python
categories = ['A', 'B', 'C']
values = [10, 20, 15]

plt.bar(categories, values )
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Bar Chart Example')
plt.grid(True)
plt.show()
```

## 9. Pie Chart

- A **pie chart** shows **parts of a whole**.
- It is useful when the number of categories is small.


```python
values = [40, 35, 25]
labels = ['Python', 'Java', 'C++']

plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title('Pie Chart Example')
plt.show()
```

## 10. Best Practices

When creating plots, students should remember:

- Add a **title**
- Add **x-axis and y-axis labels**
- Use **legend** when needed
- Use **grid** to improve readability
- Choose the correct chart type:
  - **Line plot** → trends
  - **Scatter plot** → relationships
  - **Bar chart** → category comparison
  - **Pie chart** → proportions
