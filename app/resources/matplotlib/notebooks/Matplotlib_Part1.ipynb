## Environment Setup & First Plot


Matplotlib is a Python library for creating **static, animated, and interactive visualizations**.

In most courses, we begin with `matplotlib.pyplot`, often imported as `plt`:
- Think of `plt` as a “plotting controller” that creates figures and axes and draws plots.

We'll also use the Jupyter magic `%matplotlib inline` so plots show inside the notebook.


````python
# If you're in Jupyter, this makes plots appear in the notebook
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
````

### A first minimal plot

A line plot connects points in the order given. It is best when the x-axis has a natural order (time, distance, iterations, etc.).


````python
x = np.arange(0, 10, 1)
y = x**2

plt.plot(x, y)

plt.show()
````

### Why the plot is not yet “good”

A plot is **not** complete unless it has:
- a meaningful **title**
- labeled axes (**with units if possible**)
- a legend when multiple series exist
- appropriate scales and readable ticks


## Pyplot Fundamentals (Figure, Axes, Plot)


### Key concepts

- **Figure**: the whole canvas (like a page).
- **Axes**: the area where data is plotted (you can have multiple axes in one figure).
- **Plot call**: `plt.plot(...)` draws a line on the current axes.

Even when using pyplot, it helps to *think in figure/axes terms*.


````python
# Create a figure explicitly (recommended for teaching)
plt.figure(figsize=(6, 4))
x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine wave")
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

### When to call `plt.show()`

- In scripts: usually needed to display the plot window.
- In notebooks: plots often display automatically, but using `plt.show()` is clearer and avoids surprises.


## Labels, Title, Legend, and Grid: Making Plots Understandable


### Definitions

- **Title**: what the plot is about (should answer “what?”)
- **Axis labels**: what each axis represents (with **units**)
- **Legend**: mapping from style to meaning (series names)
- **Grid**: helps reading values; use light gridlines

A plot without labels is a *picture*, not a *scientific figure*.


````python
x = np.linspace(0, 5, 200)
y1 = np.exp(-x)

plt.figure(figsize=(6, 4))

plt.plot(x, y1, label="exp(-x)")

plt.title("Exponential decay comparison")

plt.xlabel("Time (s)")

plt.ylabel("Amplitude (a.u.)")

plt.grid(True, linestyle=":")

plt.legend(title="Models")

plt.show()
````

### Common mistakes (teach students to avoid)

- Missing axis labels or units
- Using too many colors/markers
- Tiny font size
- Overplotting (too many lines/points)
- No explanation in the report text (a figure must be referenced and described)


## Line Charts: From Basic to Publication‑Ready


### 1) Plotting one series

Line charts are ideal for continuous trends.


````python
x = np.linspace(0, 10, 101)
y = np.log1p(x)   # log(1+x)

plt.figure(figsize=(7, 4))
plt.plot(x, y)
plt.title("Log growth")
plt.xlabel("x")
plt.ylabel("log(1+x)")
plt.grid(True)
plt.show()
````

### 2) Multiple lines + legend

When plotting multiple series, **always** include a legend.


````python
x = np.linspace(0, 2*np.pi, 400)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(7, 4))

plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")

plt.title("Trigonometric functions")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.grid(True)
plt.legend()
plt.show()
````

### 3) Line style, linewidth, and markers

Common formatting parameters:
- `color=` or short code `c=`
- `linestyle=` (e.g., `"-"`, `"--"`, `":"`)
- `linewidth=` (thicker for presentations)
- `marker=` (use sparingly for many points)

**Rule of thumb:** If you have many points, markers can clutter the plot.

### Common options
- `color` → changes line color
  * 'b' → blue
  * 'g' → green
  * 'r' → red
  * 'c' → cyan
  * 'm' → magenta
  * 'y' → yellow
  * 'k' → black
- `linestyle` → changes line style
  * '-' → solid
  * '--' → dashed
  * '-.' → dash-dot
  * ':' → dotted
  * '' or 'None' → no line
- `marker` → adds symbols on points
  * 'o' → circle
  * 's' → square
  * '^' → triangle up
  * 'v' → triangle down
  * '<' → triangle left
  * '>' → triangle right
  * 'x' → x mark
  * '+' → plus
  * '*' → star
  * 'd' → diamond
  * '.' → point
* `linewidth`
  * 2–3 → moderately thicker
  * 3–5 → bold, very visible (great for projectors)

````python
x = np.arange(0, 11)
y = np.array([0, 1, 1.5, 2.2, 2.9, 3.1, 3.8, 4.2, 5.1, 6.0, 6.8])

plt.figure(figsize=(7, 4))

plt.plot(x, y, c= 'k' ,linestyle=":", linewidth=1, marker="d", label="Measured")
plt.plot(x, y*2 ,linestyle="--", linewidth=1, marker=">", label="xx")



plt.title("Example with markers")
plt.xlabel("Experiment index")
plt.ylabel("Score")
plt.grid(True, linestyle=":")
plt.legend(title = 'Lines')
plt.show()
````

### 4) Limits, ticks, and readability

- Use `plt.xlim`, `plt.ylim` to focus on the important region.
- Use `plt.xticks`, `plt.yticks` to control tick placement and labels.


````python
x = np.linspace(0, 10, 500)
y = np.sin(x) + 0.1*np.random.RandomState(0).normal(size=x.size)  # Gaussian Distribution or normal distribution (around zero but with noise 0.1)

plt.figure(figsize=(7, 4))

plt.plot(x, y, label="signal")

plt.title("Noisy sine (focused view)")

plt.xlabel("x")
plt.ylabel("y")

plt.grid(True)

plt.xlim(0, 6)      # Sets the x-axis to display values only from 0 to 6

plt.ylim(-1.5, 1.5) # Sets the y-axis to display values only from -1.5 to 1.5

plt.legend()
plt.show()
````

````python
x = np.linspace(0, 10, 500)
y = np.sin(x) + 0.1*np.random.RandomState(0).normal(size=x.size)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label="signal")
plt.title("Noisy sine (focused view)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

#This chooses exactly where ticks appear
plt.xticks([0, 2, 4, 6, 8 , 10 ])
plt.yticks([-1.5, 0, 1.5])


plt.legend()
plt.show()
````

Use `plt.savefig(...)` **before** `plt.show()` for consistent saving in scripts.

## Scatter Charts: Relationships and Patterns


Scatter plots show the relationship between two variables, and are best for:
- correlation or trends
- clusters
- outliers


````python
rng = np.random.RandomState(42)
x = rng.normal(loc=0.0, scale=1.0, size=200)  # loc=0.0 → mean = 0 , scale=1.0 → standard deviation = 1, size=200 → number of values

# 0.8 * x -> A linear relationship (y depends on x)
# Random noise: mean = 0 , standard deviation = 0.6
y = 0.8*x + rng.normal(0, 0.6, size=200)

plt.figure(figsize=(6, 4))

plt.scatter(x, y, alpha=0.5 )  # alpha parameter controls transparency (opacity) of plotted elements

plt.title("Scatter plot: x vs y")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")
plt.show()
````

### Scatter with size and color encoding

You can encode extra information:
- point size `s=`
- point color `c=`
- transparency `alpha=`

This turns a 2D plot into a **multivariate** visualization.




````python
n = 300
x = rng.uniform(0, 10, n)
y = rng.uniform(0, 10, n)

sizes = 20 + 80*rng.rand(n) # small dot = 20 , large dot = 100
colors = x + y

plt.figure(figsize=(6, 5))

# default colormap (viridis): low values → dark blue/purple ,  medium → green , high values → yellow
plt.scatter(x, y, s=sizes, c=colors, alpha=0.5)

plt.title("Scatter with size & color encoding")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.colorbar(label="x + y")

plt.show()
````

`plt.scatter(x, y, c=colors, cmap="viridis")`
* "viridis" (default)
* "plasma"

````python
n = 300
x = rng.uniform(0, 10, n)
y = rng.uniform(0, 10, n)
sizes = 20 + 80*rng.rand(n) # small dot = 20 , large dot = 1
colors = x + y

plt.figure(figsize=(6, 5))

# default colormap (viridis): low values → dark blue/purple ,  medium → green , high values → yellow
plt.scatter(x, y, s=sizes, c=colors, cmap="magma", alpha=0.7)
plt.title("Scatter with size & color encoding")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")
plt.colorbar(label="x + y")
plt.show()
````

### When not to use a scatter plot

Avoid scatter plots if your data is:
- a time series (use a line plot),
- categorical on x-axis (use bar/box plots),
- extremely dense (consider hexbin / 2D histogram in advanced topics). Because:

  * scatter plots become cluttered
  * points overlap and hide patterns


## Subplots: Multiple Charts in One Figure


Subplots allow multiple visualizations in one figure.

Typical use cases:
- comparing two related signals
- showing raw data vs processed data
- showing training/validation curves side-by-side


````python
x = np.linspace(0, 2*np.pi, 400)
sinx = np.sin(x)
cosx = np.cos(x)


plt.subplot(1, 2, 1)  # rows, cols, index
plt.plot(x, sinx)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.subplot(1, 2, 2)
plt.plot(x, cosx)
plt.title("cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.suptitle("Subplots example (1 row, 2 columns)")
plt.show()
````

### Layout tip

`plt.tight_layout()` helps prevent overlapping labels/titles.

Improves spacing between subplots.


````python
x = np.linspace(0, 2*np.pi, 400)
sinx = np.sin(x)
cosx = np.cos(x)

plt.subplot(1, 2, 1)  # rows, cols, index
plt.plot(x, sinx)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.subplot(1, 2, 2)
plt.plot(x, cosx)
plt.title("cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.suptitle("Subplots example (1 row, 2 columns)")
plt.tight_layout()
plt.show()
````

````python
x = np.arange(1, 6)
y1 = x
y2 = x**2

plt.subplot(2, 1, 1)
plt.plot(x, y1, marker='o')
plt.title('First Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(x, y2, marker='s')
plt.title('Second Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.tight_layout()
plt.show()
````

plt.subplot(# rows, # column, current plot position)

Grid layout (2 rows × 2 columns):

(1) (2)

(3) (4)


````python
x = np.arange(1, 6)
y1 = x
y2 = x**2

plt.subplot(2, 2, 4)
plt.plot(x, y1, marker='o')
plt.title('First Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(x, y2, marker='s')
plt.title('Second Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.subplot(2, 2, 2)
plt.plot(x, y1, marker='o')
plt.title('3rd Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, axis="y",)

plt.subplot(2, 2, 1)
plt.plot(x, y2, marker='s')
plt.title('4th Subplot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(False)

plt.suptitle("Subplots example (2 row, 2 columns)")

plt.tight_layout()
plt.show()
````

## Bar Charts: Comparing Categories


Bar charts are used to compare **categories** or **discrete groups**.

Examples:
- number of students in each department
- average score by group
- counts by class label


````python
categories = ["A", "B", "C", "D"]
values = [23, 17, 35, 29]

plt.figure(figsize=(6, 4))

plt.bar(categories, values)

plt.title("Bar chart: Category counts")
plt.xlabel("Category")
plt.ylabel("Count")
plt.grid(True, axis="y", linestyle=":")

plt.show()
````

### Horizontal bars (useful for long labels)

````python
plt.figure(figsize=(6, 4))

plt.barh(categories, values)

plt.title("Horizontal bar chart")
plt.xlabel("Count")
plt.ylabel("Category")
plt.grid(True, axis="x", linestyle=":")
plt.show()
````

### Bar chart with annotations (values on bars)

````python
plt.figure(figsize=(6, 4))

bars = plt.bar(categories, values , color = ['red' , 'blue' , 'black', 'green'])

plt.title("Bar chart with annotations")
plt.xlabel("Category")
plt.ylabel("Count")
plt.grid(True, axis="y", linestyle=":")

for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h + 0.5, f"{h}", ha="center", va="bottom")

plt.ylim(0, max(values) + 8)
plt.show()
````

## Pie Charts: Parts of a Whole (With Caution)


Pie charts show proportions that sum to 100%.
They can be useful for simple compositions, but they are often **misused**.

**Better alternatives** for many categories:
- bar chart
- stacked bar


````python
labels = ["Rent", "Food", "Transport", "Savings"]
sizes = [40, 25, 10, 25]  # 100

plt.figure(figsize=(6, 6))

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Monthly budget composition")
plt.show()
````

````python
labels = ["Rent", "Food", "Transport", "Savings"]
sizes = [40, 25, 10, 25]

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Monthly budget composition")
plt.show()
````

# Matplotlib with Pandas

````python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500

df = pd.DataFrame({
    'ID': range(1, n + 1),
    'Math': np.random.randint(50, 100, n),
    'Physics': np.random.randint(40, 100, n),
    'Chemistry': np.random.randint(45, 100, n),
    'English': np.random.randint(50, 100, n),
    'Attendance': np.random.randint(60, 100, n)
})
# average
avg = df[["Math", "Physics", "Chemistry", "English"]].mean(axis=1)
# StudyHours depends on performance
df["StudyHours"] = ((avg / 10) + np.random.normal(0, 1, n)).clip(1, 10).astype(int)

# Save
df.to_csv('big_students.csv', index=False)

# Load again
df = pd.read_csv('big_students.csv')

print(df.head())
````

````python

avg_scores = df[['Math','Physics','Chemistry','English']].mean()

print(avg_scores)

plt.bar(avg_scores.index, avg_scores.values, color = 'red')
plt.title("Average Subject Scores")
plt.ylabel("Score")
plt.show()
````

````python
df["Pass"] = (
    (df["Math"] >= 60) &
    (df["Physics"] >= 60) &
    (df["Chemistry"] >= 60) &
    (df["English"] >= 60)
)


pass_count = df["Pass"].sum()
fail_count = len(df) - pass_count


plt.pie([pass_count, fail_count],
        labels=["Pass", "Fail"],
        autopct="%1.1f%%",
        startangle=90)

plt.title("Overall Pass vs Fail Students")
plt.show()
````

````python
plt.boxplot([
    df['Math'],
    df['Physics'],
    df['Chemistry'],
    df['English']
])

plt.title("Box Plot of Student Scores")
plt.xticks([1,2,3,4], ['Math','Physics','Chemistry','English'])
plt.ylabel("Scores")
plt.grid(True)
plt.show()
````

````python
plt.hist(df['Attendance'], color='red' , bins=20)
plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")
plt.show()
````

````python

# compute average
avg = df[["Math", "Physics", "Chemistry", "English"]].mean(axis=1)

# scatter plot
plt.scatter(df["StudyHours"], avg, alpha=0.4)

plt.title("Study Hours vs Average Grade")
plt.xlabel("Study Hours")
plt.ylabel("Average Grade")

plt.grid(True, linestyle=":")
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.RandomState(1)

# Data for line plot
x_line = np.linspace(0, 10, 200)
y_line = np.sin(x_line)

# Data for scatter plot
x_sc = rng.uniform(0, 10, 200)
y_sc = 1.5 * x_sc + rng.normal(0, 1.2, 200)

plt.figure(figsize=(12, 4))

# Left subplot: line
plt.subplot(1, 2, 1)
plt.plot(x_line, y_line, label="sin(x)")
plt.title("Task C (Left): Line Plot")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, linestyle=":")
plt.legend()

# Right subplot: scatter
plt.subplot(1, 2, 2)
plt.scatter(x_sc, y_sc, alpha=0.7)
plt.title("Task C (Right): Scatter Plot")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle=":")

plt.tight_layout()
plt.show()
````

**✅ Task D — Bar + Pie (same data) + short discussion**

````python
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D", "E"]
values = [40, 25, 20, 10, 5]

plt.figure(figsize=(12, 4))

# Bar chart
plt.subplot(1, 2, 1)
plt.bar(categories, values)
plt.title("Task D: Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.grid(True, axis="y", linestyle=":")

# Pie chart
plt.subplot(1, 2, 2)
plt.pie(values, labels=categories, autopct="%1.1f%%", startangle=90)
plt.title("Task D: Pie Chart")

plt.tight_layout()
plt.show()
````
