```python
import numpy as np

# Create two 2-D arrays (2 rows, 2 columns each)
arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[10, 20],
                 [30, 40]])

# Join along columns (axis=1)
result = np.concatenate((arr1, arr2), axis=1)

print("Array 1:")
print(arr1)
print("\nArray 2:")
print(arr2)
print("\nConcatenated along columns (axis=1):")
print(result)
```

## np.random.seed()
Seed function is used to save the state of a random function, so that it can generate same random numbers on multiple executions of the code on the same machine or on different machines (for a specific seed value). The seed value is the previous value number generated by the generator. For the first time when there is no previous value, it uses current system time.

```python
np.random.seed(10)
np.random.randint(low = 1, high = 10, size = 50)
```


## numpy.unique
- This function returns an array of unique elements in the input array.

```python
a = np.array([10,10,1,2,1,3,2,4,6,6,6])
a1=np.unique(a)
print(a1)


### u, indices= np.unique(a, return_index=True)
u, indices= np.unique(a, return_index=True)
print(u, indices)

#####

u, count1= np.unique(a, return_counts=True)
print(u, count1)

## Return the unique rows of a 2D array
aa = np.array([[1, 0, 0], [1, 0, 0], [2, 3, 4]])
a3=np.unique(aa,axis=0)
print()
print(a3)
```

## argmin and argmax

- argmax: Returns the indices of the maximum values along an axis.

```python
arr1000=np.array([1,10,-1,2,-30])
print(np.argmax(arr1000))  # unlike max it returns index

print(np.max(arr1000))
```

### numpy.argsort()

- Returns the indices that would sort an array.
- Perform an indirect sort along the given axis using the algorithm specified by the kind keyword. It returns an array of indices of the same shape as a that index data along the given axis in sorted order.

```python
# input array
in_arr = np.array([ 2, 0,  1, 5, 4, 1, 9])
#         indix=    0, 1,  2, 3, 4, 5, 6
print ("Input unsorted array : ", in_arr)

out_arr = np.argsort(in_arr)
print ("Output sorted array indices : ", out_arr)
print("Output sorted array : ", in_arr[out_arr])

## Indices lowest two values
np.argsort(in_arr)[:2]
```

```python
## np.argsort in decending order
res=np.argsort(in_arr)[::-1]
print(res)
```

#### In NumPy, you can extend or shrink the dimensions of an array using various functions and methods. Here are a few common ways to extend or shrink the axes of a NumPy array:

np.expand_dims() - Extend Axis:
The np.expand_dims() function is used to add a new axis to an array at a specified position. It takes the array and the axis position as arguments. Here's an example:

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr.shape)  # Output: (5,)

expanded_arr = np.expand_dims(arr, axis=0)
print(expanded_arr.shape)  # Output: (1, 5)
```

#### np.squeeze() - Shrink Axis:
The np.squeeze() function is used to remove single-dimensional axes from an array. It returns a new array with the same data but with the single-dimensional axes removed. Here's an example:

```python
import numpy as np

arr = np.array([[1, 2, 3, 4, 5]])
print(arr.shape)  # Output: (1, 5)

squeezed_arr = np.squeeze(arr)
print(squeezed_arr.shape)  # Output: (5,)
```
