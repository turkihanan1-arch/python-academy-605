# Python OOP

```python
# Parent class
class Person:
    def __init__(self, name="", age=0):
        self.__name = name
        self.__age = age

    # Name getter/setter
    def get_name(self):
        return self.__name

    def set_name(self, name):
        if isinstance(name, str) and name.strip() != "":
            self.__name = name
        else:
            print("Invalid name")

    # Age getter/setter
    def get_age(self):
        return self.__age

    def set_age(self, age):
        if isinstance(age, int) and age > 0:
            self.__age = age
        else:
            print("Age must be a positive integer")

    # Method to print info
    def print_info(self):
        print(f"Name: {self.__name}")
        print(f"Age : {self.__age}")


# Child class inherits from Person
class Student(Person):
    def __init__(self, name="", age=0, grade=0):
        super().__init__(name, age)
        self.__grade = grade

    # Grade getter/setter
    def get_grade(self):
        return self.__grade

    def set_grade(self, grade):
        try:
            grade = int(grade)
            if 0 <= grade <= 100:
                self.__grade = grade
            else:
                print("Grade must be between 0 and 100")
        except ValueError:
            print("Grade must be an integer")

    # Override print_info to include grade and status
    def print_info(self):
        super().print_info()
        print(f"Grade: {self.__grade}")
        print("Status:", "Passed" if self.__grade >= 60 else "Failed")
```

```python
# Create 2 student objects
student1 = Student()
student2 = Student()

students = {"1": student1, "2": student2}

while True:
    print("\n--- Main Menu ---")
    print("1. Print All Students Info")
    print("2. Select Student to Update")
    print("3. Exit")

    main_choice = input("Enter your choice (1-3): ")

    if main_choice == "3":
        print("Exiting program. Goodbye!")
        break

    elif main_choice == "1":
        print("\n--- All Students Info ---")
        for key, student in students.items():
            print(f"\nStudent {key}:")
            student.print_info()

    elif main_choice == "2":
        print("\n--- Select Student ---")
        print("1. Student 1")
        print("2. Student 2")
        s_choice = input("Enter your choice (1-2): ")

        if s_choice not in ["1", "2"]:
            print("Invalid choice! Enter 1 or 2.")
            continue

        student = students[s_choice]

        # Menu for the selected student
        while True:
            print(f"\n--- Menu for Student {s_choice} ---")
            print("1. Enter/Update Name")
            print("2. Enter/Update Age")
            print("3. Enter/Update Grade")
            print("4. Back to Main Menu")

            choice = input("Enter your choice (1-4): ")

            if choice == "1":
                student.set_name(input("Enter Name: "))
            elif choice == "2":
                try:
                    student.set_age(int(input("Enter Age: ")))
                except ValueError:
                    print("Please enter a valid integer for age")
            elif choice == "3":
                student.set_grade(input("Enter Grade: "))
            elif choice == "4":
                break
            else:
                print("Invalid choice! Enter a number between 1 and 5.")

    else:
        print("Invalid choice! Enter a number between 1 and 3.")
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Manager(Employee):
    def add_bonus(self, bonus):
        self.total_salary = self.salary + bonus

    def display(self):
        print("Name =", self.name)
        print("Total Salary =", self.total_salary)

m = Manager("YourName", 30000)
m.add_bonus(5000)
m.display()
```
