### File Handling
The key function for working with files in Python is the open() function.

The open() function takes two parameters; filename, and mode.

There are four different methods (modes) for opening a file:
*   "r" - Read - Default value. Opens a file for reading, error if the file does not exist
*   "a" - Append - Opens a file for appending, creates the file if it does not exist
*   "w" - Write - Opens a file for writing, creates the file if it does not exist
*   "x" - Create - Creates the specified file, returns an error if the file exists
In addition you can specify if the file should be handled as binary or text mode
*   "t" - Text - Default value. Text mode
*   "b" - Binary - Binary mode (e.g. images)

### Syntax
To open a file for reading it is enough to specify the name of the file:

`f = open("demofile.txt")`

The code above is the same as:

`f = open("demofile.txt", "rt")`

Because "r" for read, and "t" for text are the default values, you do not need to specify them.

Create a New File
To create a new file in Python, use the open() method, with one of the following parameters:
*   "x" - Create - will create a file, returns an error if the file exists

*   "w" - Write - will create a file if the specified file does not exists

````python
file1 = open("myfile.txt", "a")

# Result: a new empty file is created.
````

<font color="blue" size =5>2. Python File Open</font>


Assume we have the following file, located in the same folder as Python:


```
Hello! Welcome to myfile.txt
This file is for testing purposes.
Good Luck!
```

To open the file, use the built-in open() function.

The open() function returns a file object, which has a read() method for reading the content of the file:


````python
f = open("myfile.txt")
print(f.read())
````

If the file is located in a different location, you will have to specify the file path, like this:


```
f = open("D:\\myfiles\welcome.txt")
print(f.read())
```

<font color="blue" size =5>3. Close Files</font>

It is a good practice to always close the file when you are done with it.

If you are not using the with statement, you must write a close statement in order to close the file:

````python
try:
  f = open("myfile.txt")
  txt = f.readline()
  print(txt.upper() )

except:
  pass
finally:
  f.close()
````



---


### <font color="blue" size =5>Using the with statement</font>

You can also use the with statement when opening a file:

````python
with open("myfile.txt") as f:
  txt = f.readline()
  print(txt.upper() )

````

Then you do not have to worry about closing your files, the with statement takes care of that.

### <font color="blue" size =5>4. Write to an Existing File</font>


To write to an existing file, you must add a parameter to the open() function:
*   "a" - Append - will append to the end of the file
*   "w" - Write - will overwrite any existing content

````python
'''
ExampleGet your own Python Server
Open the file "myfile.txt" and append content to the file:
'''
with open("myfile.txt", "a") as f:
  f.write("Now the file has more content!\n")

#open and read the file after the appending:
with open("myfile.txt") as f:
  print(f.read())
````

Overwrite Existing Content
To overwrite the existing content to the file, use the w parameter:

````python
'''
Example
Open the file "myfile.txt" and overwrite the content:
'''

with open("myfile.txt", "w") as f:
  f.write("Woops! I have deleted the content!\n")

#open and read the file after the overwriting:
with open("myfile.txt") as f:
  print(f.read())
````

### <font color="blue" size =5>5. Delete a File</font>


To delete a file, you must import the OS module, and run its os.remove() function:

````python
'''
ExampleGet your own Python Server
Remove the file "myfile.txt":
'''

import os
os.remove("myfile.txt")
````

### <font color="blue" size =5>Check if File exist:</font>


To avoid getting an error, you might want to check if the file exists before you try to delete it:

````python
'''
Example
Check if file exists, then delete it:
'''

import os
if os.path.exists("testFile.txt"):
  print("File exist")
  os.remove("testFile.txt")
  print("file Removed")
else:
  print("The file does not exist")
````
