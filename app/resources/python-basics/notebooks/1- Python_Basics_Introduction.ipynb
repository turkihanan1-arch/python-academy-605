# Compiler vs Interpreter
- Programming languages require a compiler or interpreter to translate code into instructions that the computer can execute.

- Compiler: Converts entire code into machine code before program runs. Advantages are faster program execution. Disadvantage is compile time before run. Examples are C, C++, Rust.

- Interpreter: Converts code line-by-line as program runs. Advantage is no compile time needed. Disadvantage is slower execution. Examples are Python, JavaScript, Ruby.

- Python uses an interpreter which allows for rapid development cycles without compile time.

## Reasons for using Jupyter ?


Jupyter Notebooks are widely used in the data science and scientific computing communities for several compelling reasons:

1. **Interactive Computing**: Jupyter allows for interactive code execution, which means you can run bits of code in separate cells independently. This is particularly useful for data exploration, experimentation, and iterative coding.

2. **Documentation**: You can document your code, its execution, and the results in a single document using Markdown, which supports rich text formatting, images, hyperlinks, and even LaTeX for mathematical expressions.

3. **Data Visualization**: Jupyter supports inline visualization with libraries like Matplotlib, Seaborn, and others. Graphs and plots can be displayed directly below the code cells that produce them, making it easy to analyze and iterate on visual data.

4. **Support for Many Programming Languages**: Although it was initially designed for Python, Jupyter now supports over 40 programming languages, including R, Julia, and Scala, through the use of different kernels.

5. **Reproducibility**: Jupyter Notebooks can be shared with others, allowing them to see the code, the output, and additional comments and explanations, which makes it easier to reproduce the work and verify the results.

6. **Educational Purposes**: They are a great tool for teaching coding and computational thinking because they allow the creation of a document with live code, visualizations, and narrative text.

7. **Integration with Data Science Tools**: Jupyter integrates with big data processing tools like Apache Spark, and it can interface with databases, which makes it a powerful tool for querying and analyzing large datasets.

8. **Extension and Widgets**: Jupyter has a growing ecosystem of extensions that add functionality to notebooks. This includes widgets that can create interactive UI elements for live data manipulation.

9. **Flexible and Portable**: Jupyter Notebooks are saved as JSON files with a `.ipynb` extension. They can be version controlled, converted to other formats such as HTML or PDF, and run on different machines or in the cloud.

10. **Development and Debugging**: It provides a convenient environment for developing and debugging code snippets, testing hypotheses, or learning new libraries by allowing you to execute small portions of code independently.

These features make Jupyter Notebooks a valuable tool for a wide range of tasks, from beginner's learning to code to researchers conducting complex data analysis.
